```python
# Part 1

def f(x):
    """Calculate the value of the function f(x) = x^3 - x^2 - 1.

    Parameters: x - A numerical value for which to evaluate the function f(x).
    Returns: The value of f(x) for the given input x.
"""
    return x**3 - x**2 - 1

def df(x):
    """Calculate the value of the derivative of the function f(x) = x^3 - x^2 - 1.
    Parameters: x - A numerical value for which to evaluate the derivative df(x).
    Returns: The value of df(x) for the given input x.
    """
    return 3*x**2 - 2*x
```

```python
# part 2
import numpy as np
def newton(f, df, x0, epsilon=1e-6, max_iter=30):
    """Use the Newton iteration method on a function to find
    its root
    
    Parameters:
    f: function to find root of
    df: derivative of function
    x0: Initial x to start iterating on
    epsilon: Value that when |f(xn)| is less than, the root is returned
    max_iter: The maximum number of iterations 
    
    Returns:
    xn: The root closest to x0 such that f(xn) = 0"""

    x_n = x0
    greater_than_e = np.abs(f(x_n)) > epsilon
    i = 0
    while greater_than_e == True and i < max_iter:
        x_n1 = x_n - f(x_n)/df(x_n)
        x_n = x_n1
        greater_than_e = np.abs(f(x_n)) > epsilon

        i+=1

    if not greater_than_e:
        print(f"Found root in {i} iterations.")
        return x_n
    
    else:
        print(f"Iteration failed")
        return None
```

```python
# Part 3

x_n_ex1 = newton(f, df, 1)
print(f"closest root is {x_n_ex1}")
```

```python
x_n_ex2 = newton(f, df, -200)
print(f"closest root is {x_n_ex2}")
```

```python
x_n_smaller_ex1 = newton(f, df, 1, epsilon = 1e-8)
print(f"closest root is {x_n_smaller_ex1}")
```
